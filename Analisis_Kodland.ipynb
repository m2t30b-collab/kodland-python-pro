## Datos originales
Se conservan los 1.000 registros sin corregir ni inventar valores.

```python
from google.colab import files
uploaded = files.upload()
df = pd.read_csv('kodland_data.csv')
assert len(df) == 1000
```

## Cinco respuestas

```python
minimo = int(df.points.min())
maximo_tareas = int(df.homework.max())
menores = df.loc[df.points == minimo, ['id', 'course']]
sin_tareas = int((df.homework == 0).sum())
puntos_python_pro = int(df.loc[df.course == 'Python Pro', 'points'].sum())
respuestas = [
    f'Tarea 1: minimo de puntos = {minimo}',
    f'Tarea 2: maximo de deberes = {maximo_tareas} %',
    'Tarea 3: alumnos con menos puntos:\n' + menores.to_string(index=False),
    f'Tarea 4: alumnos que nunca hicieron deberes = {sin_tareas}',
    f'Tarea 5: suma de puntos de Python Pro = {puntos_python_pro}',
]
print('\n\n'.join(respuestas))
Path('answers.txt').write_text('\n\n'.join(respuestas), encoding='utf-8')
```
